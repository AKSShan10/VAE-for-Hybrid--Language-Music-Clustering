```python
# Example of running the model and clustering in a Jupyter Notebook
import numpy as np
from src.dataset import extract_audio_features, extract_lyrics_features
from src.vae import vae
from src.clustering import kmeans_clustering, visualize_clusters_with_tsne
from src.evaluation import evaluate_clustering

# Load and preprocess the dataset (audio + lyrics)
audio_files = ['audio_file_1.wav', 'audio_file_2.wav', 'audio_file_3.wav']
lyrics_files = ['lyrics_file_1.txt', 'lyrics_file_2.txt', 'lyrics_file_3.txt']

# Extract features
audio_features = extract_audio_features(audio_files)
lyrics_features = extract_lyrics_features(lyrics_files)

# Combine audio and lyrics features
combined_features = np.concatenate([audio_features, lyrics_features], axis=-1)

# Train the VAE on combined features
vae.fit(combined_features, combined_features, epochs=50, batch_size=32)

# Extract latent features from VAE encoder
latent_features = vae.encoder.predict(combined_features)[0]

# Perform KMeans clustering
kmeans_labels = kmeans_clustering(latent_features, n_clusters=3)

# Visualize the clusters
visualize_clusters_with_tsne(latent_features, kmeans_labels)

# Evaluate clustering performance
evaluate_clustering(latent_features, kmeans_labels)
```
