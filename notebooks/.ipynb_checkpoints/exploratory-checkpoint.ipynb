Cell 2 — Helper utilities (saving plots + showing saved plots)

```python
from pathlib import Path
from PIL import Image

def ensure_dir(p: str):
    os.makedirs(p, exist_ok=True)

def savefig(path: str, dpi=200):
    ensure_dir(os.path.dirname(path))
    plt.tight_layout()
    plt.savefig(path, dpi=dpi)
    plt.close()
    print("Saved figure:", path)

def show_image(path: str, title: str | None = None, max_size=(1000, 600)):
    if not os.path.exists(path):
        print("Missing:", path)
        return
    img = Image.open(path)
    img.thumbnail(max_size)
    plt.figure()
    plt.imshow(img)
    plt.axis("off")
    if title:
        plt.title(title)
    plt.show()

def list_saved_results(pattern="../results/**/*.png"):
    paths = sorted(glob.glob(pattern, recursive=True))
    print("Found", len(paths), "png files")
    for p in paths[:30]:
        print(" -", p)
    if len(paths) > 30:
        print(" ...")
    return paths
```

2D embedding of Hybrid latent (UMAP/t-SNE) and save plots per clustering method

```python
from sklearn.manifold import TSNE

try:
    import umap
    HAS_UMAP = True
except Exception:
    HAS_UMAP = False

def embed_2d(Z, seed=42):
    if HAS_UMAP:
        return umap.UMAP(n_neighbors=15, min_dist=0.1, random_state=seed).fit_transform(Z)
    perplexity = min(30, max(5, Z.shape[0] // 20))
    return TSNE(n_components=2, init="pca", random_state=seed, perplexity=perplexity).fit_transform(Z)

Z2 = embed_2d(Z_hybrid, seed=42)
method_name = "umap" if HAS_UMAP else "tsne"

for m, labels in labels_by_method.items():
    plt.figure(figsize=(7, 5))
    plt.scatter(Z2[:, 0], Z2[:, 1], c=labels, s=10)
    plt.title(f"Medium Task: Hybrid Latent 2D ({method_name.upper()}) - {m}")
    plt.tight_layout()
    out_path = os.path.join(RESULTS_DIR, "medium_visualization", f"hybrid_{method_name}_{m}.png")
    plt.savefig(out_path, dpi=200)
    plt.close()
    print("Saved:", out_path)

# Show one plot per method
for m in labels_by_method:
    show_image(os.path.join(RESULTS_DIR, "medium_visualization", f"hybrid_{method_name}_{m}.png"),
               f"Hybrid 2D - {m}")
```

```python
import chardet
from sklearn.feature_extraction.text import TfidfVectorizer
import os

def extract_lyrics_features(lyrics_dir, max_features=5000):
    """
    Extracts TF-IDF features from the lyrics with automatic encoding detection and fallback.
    """
    lyrics = []
    
    # List of encodings to try in case of failure
    encodings_to_try = ['utf-8', 'gb2312', 'big5', 'cp949', 'utf-16']
    
    for lyric_file in os.listdir(lyrics_dir):
        try:
            lyrics_text = ""
            for encoding in encodings_to_try:
                try:
                    # Attempt to read the file using each encoding
                    with open(os.path.join(lyrics_dir, lyric_file), 'r', encoding=encoding) as file:
                        lyrics_text = file.read()
                        print(f"Successfully read {lyric_file} using {encoding} encoding.")
                        break  # Stop when we successfully read the file
                except UnicodeDecodeError:
                    # If one encoding fails, try the next one
                    continue
            
            # If no encoding worked, use the 'replace' error handling strategy
            if not lyrics_text:
                with open(os.path.join(lyrics_dir, lyric_file), 'r', encoding='utf-8', errors='replace') as file:
                    lyrics_text = file.read()
                print(f"Failed with all encodings, using 'utf-8' with errors='replace' for {lyric_file}.")
            
            lyrics.append(lyrics_text)
        
        except Exception as e:
            print(f"Error reading {lyric_file}: {e}")
            continue
    
    # Use TF-IDF to convert text data into numerical features
    tfidf = TfidfVectorizer(max_features=max_features, stop_words='english', ngram_range=(1, 2), min_df=1)
    lyrics_features = tfidf.fit_transform(lyrics).toarray()

    return lyrics_features


```

```python
# Step 2: Extract Lyrics Features (TF-IDF embeddings)
lyrics_features = extract_lyrics_features(LYRICS_DIR)
```

```python
# Step 1: Extract Audio Features (log-mel spectrograms or MFCCs)
audio_features, ids = extract_audio_features(AUDIO_DIR)
```

```python
# Before training, normalize audio features to [0, 1] range
audio_features_min = audio_features.min()
audio_features_max = audio_features.max()
audio_features_normalized = (audio_features - audio_features_min) / (audio_features_max - audio_features_min + 1e-8)

# Alternatively, use standardization (mean=0, std=1)
# audio_features_normalized = (audio_features - audio_features.mean()) / (audio_features.std() + 1e-8)

# Normalize lyrics features as well
lyrics_features_min = lyrics_features.min()
lyrics_features_max = lyrics_features.max()
lyrics_features_normalized = (lyrics_features - lyrics_features_min) / (lyrics_features_max - lyrics_features_min + 1e-8)

# Check for NaN in inputs
print(f"Audio NaN count: {np.isnan(audio_features_normalized).sum()}")
print(f"Lyrics NaN count: {np.isnan(lyrics_features_normalized).sum()}")
print(f"Audio range: [{audio_features_normalized.min()}, {audio_features_normalized.max()}]")
print(f"Lyrics range: [{lyrics_features_normalized.min()}, {lyrics_features_normalized.max()}]")
```

```python
# Step 3: Build and Compile the Conditional VAE
latent_dim = 32
epochs = 150
beta = 0.5
```

```python
# Import necessary libraries
from src.vae_conditional import build_conditional_vae  # Import the CVAE model
from src.clustering import multi_modal_clustering
import tensorflow as tf
```

```python
# Build model
encoder, decoder, cvae = build_conditional_vae(
    input_shape_audio=audio_features.shape[1:], 
    input_shape_lyrics=lyrics_features.shape[1:], 
    latent_dim=latent_dim, 
    beta=beta
)

# Use lower learning rate with learning rate scheduler
lr_schedule = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=1e-4,
    decay_steps=1000,
    decay_rate=0.96
)

cvae.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=lr_schedule))

# Train with normalized data
history = cvae.fit(
    [audio_features_normalized, lyrics_features_normalized], 
    epochs=epochs, 
    batch_size=32, 
    verbose=1,
    validation_split=0.1  # Add validation to monitor overfitting
)
```
